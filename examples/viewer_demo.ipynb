# islide — viewer demo

The canvas-based viewer. `islide.SlideViewer` is a custom Jupyter widget:
Python owns the slide (open, level selection, one-read-per-viewport, tile
cache, JPEG encode); the JS `SlideView` owns the canvas, the pan/zoom
transform, and the mouse. The minimap, zoom/fit/1:1 toolbar, and the
`status` line are part of the view.

> **Note:** the interactive canvas requires the JupyterLab/notebook
> extension (`frontend/`) to be built and installed. In environments
> without it the cells below still run (open + render happen in the
> kernel) but draw no canvas.

```
from islide import SlideViewer

# Test slide (data/ is gitignored): download data/CMU-1.tiff from
# https://openslide.cs.cmu.edu/download/openslide-testdata/Generic-TIFF/CMU-1.tiff
v = SlideViewer("data/CMU-1.tiff")   # opens the slide (synchronous)
v
```

The slide opens **synchronously in the constructor** (opening a large WSI
can take a few seconds, so the cell blocks until it is ready; a bad path
raises). Because the open is done by the time this cell runs, the widget's
first sync to the frontend already carries the complete state (open slide,
fit viewport, initial tiles). Once ready, inspect the synced metadata:

```
v.meta
```

Navigate with the mouse: **scroll** to zoom at the cursor, **drag** to
pan, or use the **minimap** and the `fit / 1:1` toolbar. Pan/zoom is
applied instantly in the view
(local transform) and the new viewport is synced back to Python (debounced),
which then fetches any missing tiles.

## Tiles & geometry

`v.tiles` maps `"level:tx:ty"` → JPEG data URL; `v.tile_geo` maps the same
key → `[level, ox, oy, cw, ch]` (absolute level-pixel crop origin + size).
The view reprojects this geometry under its local transform, so panning
stays smooth between Python round-trips.

```
import json
keys = list(v.tiles)
print(f"{len(keys)} tiles for the current viewport")
for k in keys[:4]:
    print(f"  {k}: geo={v.tile_geo[k]}  url_len={len(v.tiles[k])}")
```

## Rendering feel

This section is **feel**, not features — the wire gains one small trait (`resync`, the attach re-render counter) and the `tiles` / `tile_geo` push becomes per-chunk (module version 2.1.0). Two things change when you interact:

1. **Center-first chunked reads** (Python). A render's read plan is split
   into 1024-px blocks (4×4 tiles), each fetched independently, ordered by distance to the viewport center, and **pushed one chunk at a time** in that order — the center block lands on the canvas without waiting for the rest of the pass; a view that fits one block renders exactly as before (one read, one push, byte-identical `tiles` / `tile_geo`).
2. **Level cross-fade** (JS only). When the selected pyramid level changes
   between pushes, the old level fades out over 300 ms
   (`frontend/blend.js`, `BLEND_MS`) while the new level draws at full
   opacity. The view accumulates tile geometry across pushes (`_tileGeo`),
   so the successive pushes of a per-chunk render never blank the canvas, and
   the outgoing level stays visible until its alpha reaches zero.


```
meta = v.backend.meta
print(f"{meta.dimensions[0]}x{meta.dimensions[1]}, "
      f"{meta.level_count} levels, ds={meta.level_downsamples}")
```

## The cross-fade

In the canvas view, quickly scroll-zoom from the whole-slide view (a
coarse level) down to high magnification (level 0) — e.g. a few scroll steps, or a drag on the minimap. As the selected pyramid
level changes, the *old* level does not pop off: it fades out over
`BLEND_MS = 300 ms` while the new level draws at full opacity
(`frontend/blend.js`). The fade is computed in the compositor per level
(`levelAlphas`), coarsest level first, so the fine level is always on top
while the coarse one dissolves underneath.

Note what does **not** happen: the canvas never blanks between the
pushes of a per-chunk render (the view draws everything accumulated in
`_tileGeo`, not just the latest push), and at rest there is no animation
loop — the rAF stepping stops when `blend.done(...)` turns true.


## The chunked read plan (pure math, no pixels)

`plan_viewport` now carries the tile set **and** its chunking: 1024-px
blocks, grid-anchored, center-first. At a wide viewport the plan spans
several blocks; the order below is the fetch order — and, for a
multi-chunk render, the first block is the one pushed before the rest.

```
from islide.plan import plan_viewport
from islide.viewport import Viewport

cx, cy = meta.dimensions[0] / 2, meta.dimensions[1] / 2


def show_plan(zoom: float, dx: float = 0, dy: float = 0) -> None:
    vp = Viewport(cx + dx, cy + dy, zoom, canvas_w=960, canvas_h=540)
    plan = plan_viewport(meta, vp, tile_size=256)
    print(f"zoom {zoom:>6}: level {plan.level}, "
          f"{len(plan.tiles)} tiles in {len(plan.chunks)} chunk(s)")
    for i, c in enumerate(plan.chunks):
        print(f"  chunk {i}: origin {c.read_origin}, size {c.size}, "
              f"{len(c.tiles)} tiles")


# wide view: the canvas spans several 1024-px blocks -> multi-chunk plan
show_plan(0.5)
# tight view: one block -> one read, one push
# (offset 100 px from the exact center so the view sits inside one
# 1024-px block instead of on its corner)
show_plan(16.0, dx=100, dy=100)
```

## The per-chunk push

Set up an observer on `tiles` before zooming, then change the viewport
in the view (or via `v.set_zoom`). A wide, multi-chunk viewport produces
**one `tiles` push per chunk** — the viewport-center chunk first, the rest in center-out order — while a single-chunk viewport produces exactly one. Every push carries the same absolute `tile_geo` geometry (each tile's crop is read once per render). The traits are last-event slots — they hold the last chunk pushed, and the view accumulates a render's pushes — which is why the view bumps the `resync` counter on attach and the kernel re-renders the current viewport.

```
pushes = []
v.observe(lambda c: pushes.append(c["new"]), names="tiles")

# wide view (multi-chunk): expect two pushes, first is the center chunk
v.set_zoom(0.5, cx=meta.dimensions[0] / 2, cy=meta.dimensions[1] / 2)
print(f"pushes after wide view: {len(pushes)}")
if len(pushes) >= 2:
    print(f"  first push: {len(pushes[0])} tiles (center chunk)")
    print(f"  full set:   {len(pushes[-1])} tiles")

# tight view (single chunk): expect exactly one push, one read
# (offset 100 px from the exact center so the view is well inside
# one 1024-px block, not on a block boundary)
pushes.clear()
v.set_zoom(16.0, cx=meta.dimensions[0] / 2 + 100, cy=meta.dimensions[1] / 2 + 100)
print(f"pushes after tight view: {len(pushes)} "
      f"({len(pushes[0]) if pushes else 0} tiles)")
```

## What this is not

- Not multi-level *serving*: the view still shows one level at a time;
  the second level is only visible *during* the 300 ms fade.
- Not a tile cache change: tiles are cached by key as before; chunks
  only change how reads are grouped and in which order they arrive.
- Not a wire change: `tiles`, `tile_geo`, `minimap_img`, the overlay
  traits, and the annotation contract are all unchanged.
